## Используем встроенное неявное ожидание, которые применяется ВЕЗДЕ

```python
from selenium.webdriver.chrome.webdriver import WebDriver
from selenium.webdriver.common.by import By

try:
    driver = WebDriver()
    
    # IMPLICITLY WAIT
    driver.implicitly_wait(10)
    
    driver.get("https://xn----btbcgdcu9bgbc3ak6fzb.online/")
    
    enter_button = driver.find_element(By.ID, "enter")
    enter_button.click()
    
    phone_input = driver.find_element(By.NAME, "phone")
    phone_input.send_keys("+79061551902")
    
    password_input = driver.find_element(By.NAME, "password")
    password_input.send_keys("123456")
    
    login_button = driver.find_element(
        By.XPATH, ".//input[@type='submit' and @name='btn_submit_auth']"
    )
    login_button.click()
    
    error_message_block = driver.find_element(
        By.XPATH, ".//form[@name='form_auth']//b[@style='color:darkred']"
    )
    
    error_text = error_message_block.text
    print(error_text == "Неправильный номер и/или пароль")
    
finally:
    driver.quit()
```

## Теперь тоже самое с явными ожданиями

```python
from selenium.webdriver.chrome.webdriver import WebDriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

try:
    driver = WebDriver()
    wait = WebDriverWait(driver, timeout=10)
    
    driver.get("https://xn----btbcgdcu9bgbc3ak6fzb.online/")
    
    enter_button = wait.until(
        EC.element_to_be_clickable((By.ID, "enter"))
    )
    enter_button.click()
    
    phone_input = wait.until(
        EC.presence_of_element_located((By.NAME, "phone"))
    )
    phone_input.send_keys("+79061551902")
    
    password_input = wait.until(
        EC.presence_of_element_located((By.NAME, "password"))
    )
    password_input.send_keys("123456")
    
    login_button = wait.until(
        EC.element_to_be_clickable(
            (By.XPATH, ".//input[@type='submit' and @name='btn_submit_auth']")
        )
    )
    login_button.click()
    
    error_message_block = wait.until(
        EC.visibility_of_element_located(
            (By.XPATH, ".//form[@name='form_auth']//b[@style='color:darkred']")
        )
    )
    
    error_text = error_message_block.text
    print(error_text == "Неправильный номер и/или пароль")
    
finally:
    driver.quit()
```

```python
def burger_is_visible(driver: WebDriver):
    try:
        btn = driver.find_element(By.ID, "navbarCollapse")
        if btn.get_dom_attribute("class") == 'region region-primary-menu navbar-collapse collapse show':
            return btn
        return False
    except:
        return False
```
