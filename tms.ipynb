```python
def create_tester(url: str, name: str, grade: int) -> dict:
    tester_id = requests.post(f'{url}/testers', json={'name': name, 'grade': grade}).json()
    response = requests.get(f'{url}/testers?id={tester_id}')
    data = response.json()
    print(data)
    return data
```

```python
create_tester(DEFAULT_URL, 'ivan', 1)
```
